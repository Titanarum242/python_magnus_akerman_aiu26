```python
import random as rnd
count_sixes=0
number=1000000
for i in range(number):
    #print(f"{rnd.randint(1,6)}", end=", ")
    dice = rnd.randint(1,6)
    if dice==6:
        count_sixes +=1

(count_sixes/number)-1/6
for i in range(10):
    result=(count_sixes/number)-1/6
    print(f"{result}")
```

```python
for row in range(3):
    for col in range(3):
        for z in range(1):
            print(f"({row},{col},{z})", end=" ")
    print()        
        
```
